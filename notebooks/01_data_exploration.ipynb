# 01 – AIS Data Exploration

This notebook loads the sample AIS dataset, runs the preprocessing pipeline, and explores the data through summary statistics and visualisations.


```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_preprocessing import load_ais_csv, preprocess, data_summary

sns.set_theme(style='whitegrid')
%matplotlib inline
```

## Load & Preprocess

```python
raw = load_ais_csv('../data/sample/ais_sample.csv')
print(f'Raw shape: {raw.shape}')
raw.head()
```

```python
df = preprocess(raw)
print(f'Clean shape: {df.shape}')
data_summary(df)
```

## Vessel & Destination Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

vessel_counts = df.groupby('mmsi').size()
axes[0].hist(vessel_counts, bins=20, edgecolor='black')
axes[0].set_xlabel('Reports per vessel')
axes[0].set_ylabel('Count')
axes[0].set_title('Trajectory Length Distribution')

dest_counts = df['destination'].value_counts()
dest_counts.plot(kind='bar', ax=axes[1])
axes[1].set_title('Destination Frequency')
axes[1].set_xlabel('Destination')
axes[1].set_ylabel('Num. reports')
plt.tight_layout()
plt.show()
```

## Speed & Course Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['sog'].hist(bins=40, ax=axes[0], edgecolor='black')
axes[0].set_title('Speed Over Ground (SOG) Distribution')
axes[0].set_xlabel('SOG (knots)')

df['cog'].hist(bins=36, ax=axes[1], edgecolor='black')
axes[1].set_title('Course Over Ground (COG) Distribution')
axes[1].set_xlabel('COG (degrees)')
plt.tight_layout()
plt.show()
```

## Spatial Coverage (Scatter)

```python
plt.figure(figsize=(10, 6))
for dest, grp in df.groupby('destination'):
    plt.scatter(grp['lon'], grp['lat'], s=5, alpha=0.5, label=dest)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Vessel Positions by Destination')
plt.legend(markerscale=3, fontsize=8)
plt.tight_layout()
plt.show()
```
