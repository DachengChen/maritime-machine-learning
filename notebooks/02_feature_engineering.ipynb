# 02 – Feature Engineering

Demonstrates the full feature engineering pipeline and visualises key engineered features.


```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_preprocessing import load_ais_csv, preprocess
from src.feature_engineering import build_features, get_feature_matrix, DEFAULT_FEATURE_COLS

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
raw = load_ais_csv('../data/sample/ais_sample.csv')
df = preprocess(raw)
df_feat = build_features(df)
print(f'Shape after feature engineering: {df_feat.shape}')
df_feat.head()
```

## Temporal Encoding

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, (sin_col, cos_col, title) in zip(axes, [
    ('hour_sin', 'hour_cos', 'Hour of Day'),
    ('dow_sin',  'dow_cos',  'Day of Week'),
    ('month_sin','month_cos','Month'),
]):
    ax.scatter(df_feat[sin_col], df_feat[cos_col], alpha=0.2, s=8)
    ax.set_xlabel(f'{title} – sin')
    ax.set_ylabel(f'{title} – cos')
    ax.set_title(f'Cyclical Encoding: {title}')
    ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```

## Kinematic Features

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df_feat['delta_sog'].dropna().hist(bins=40, ax=axes[0], edgecolor='black')
axes[0].set_title('Delta SOG Distribution')
axes[0].set_xlabel('ΔSOG (knots)')

df_feat['distance_nm'].dropna().hist(bins=40, ax=axes[1], edgecolor='black')
axes[1].set_title('Step Distance Distribution')
axes[1].set_xlabel('Distance (nm)')
plt.tight_layout()
plt.show()
```

## Feature Correlation Matrix

```python
X = get_feature_matrix(df_feat, dropna=True)
corr = X.corr()

plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap='coolwarm', center=0, annot=False, linewidths=0.3)
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
